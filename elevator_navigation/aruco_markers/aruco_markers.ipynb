```python
import cv2
import numpy as np

floor = 6  # Change this to the desired floor number (0-49 for DICT_4X4_50)
# Select the dictionary you used in your ROS2 code
dictionary = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_50)

# Generate Marker ID 4
# 800 is the pixel size; high res is better for clear printing
marker_image = cv2.aruco.generateImageMarker(dictionary, floor, 800)

# Save it to a file
cv2.imwrite(f"marker_floor_{floor}.png", marker_image)
print(f"Marker generated as marker_floor_{floor}.png")
```
